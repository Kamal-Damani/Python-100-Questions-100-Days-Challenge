```python
# Q2. Leap year or Not
# logic year / 4 =0 and year /100 !=0 or year/400 =0

year = int(input("Enter the year : " ))
if 1000 <= year <= 9999:
  if (year%4 == 0 and year%100 != 0 ) or year%400 ==0:
    print(year , "is leap year")
  else:
    print(year,"is not leap year")
else :
  print("Enter valid Year")
```

```python
# Q3. Reverse of a number

def reverse(a):
    r = 0

    while a > 0:
        digit = a % 10
        r = r * 10 + digit
        a = a // 10

    return r


a = int(input("Enter number: "))
print(reverse(a))
```

```python
# Q5. write a program to find first 25 odd number

count= 0
for i in range(1,100):
  if i%2!=0:
    print(i)
    count+=1

    if count==25:
      break
```
